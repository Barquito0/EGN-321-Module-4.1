## Verified classroom calculation

```python
def estimate_minutes(batch_checks, cycle_seconds, stations, efficiency=1.0):
    return batch_checks * cycle_seconds / (stations * efficiency * 60.0)

for stations in range(1, 6):
    print(stations, estimate_minutes(600, 30, stations))
```
